```python
import json, sys
from pathlib import Path
import numpy as np, pandas as pd
from IPython.display import Image, display, Markdown
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
R, F = ROOT / "results", ROOT / "figures"
pd.set_option("display.precision", 3); pd.set_option("display.width", 200)
def fig(name):
    p = F / f"{name}.png"
    display(Image(str(p))) if p.exists() else print("missing figure:", p.name)
def csv(name):
    p = R / name
    return pd.read_csv(p) if p.exists() else print("missing table:", p.name)
```

```python
for bb in ['vit_s16', 'resnet50']:
    s = csv(f'ctta_{bb}_summary.csv')
    if s is not None:
        display(Markdown(f'**{bb}**'))
        display(s[['sequence','method','online_mean_mean','online_mean_std','final_mean_mean','bwt_mean','forgetting_source_mean','online_round1_mean','online_round2_mean']])
```

```python
for bb in ['vit_s16', 'resnet50']:
    for seq in ['brief', 'year', 'brief_long']:
        for n in ['online', 'source', 'matrix']:
            fig(f'07_ctta_{bb}_{seq}_{n}')
    fig(f'07_ctta_{bb}_detector')
```
